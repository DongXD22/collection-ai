```python
import numpy as np
```

- 数据生成

```python
point_A=np.random.randint(0,101,size=[5,2])
point_B=np.random.randint(0,101,size=[8,2])
```

- 距离矩阵计算

```python
diff=point_A[:,np.newaxis,:] - point_B[np.newaxis,:,:]
```

```python
distance_matrix=np.sqrt(np.sum(diff**2, axis=2))
```

```python
distance_matrix
```

- 轴向数据聚合

```python
np.min(distance_matrix,axis=1)
```

- 复合条件查询

```python
np.where(np.any(distance_matrix.T<20,axis=1))
```
