```python
import matplotlib.pyplot as plt
import math
import numpy as np
```

```python
def logistic(x:int,L:int=1,k:int=1,x0:int=0):
    return L/(1+(math.e)**(-k*(x-x0)))
```

```python
x=np.linspace(-10,10,100)
```

```python
plt.subplot(2,2,1)
plt.plot(x,logistic(x))
plt.title('ori')

plt.subplot(2,2,2)
plt.plot(x,logistic(x,L=3))
plt.title('L=3')

plt.subplot(2,2,3)
plt.plot(x,logistic(x,k=3))
plt.title('k=3')

plt.subplot(2,2,4)
plt.plot(x,logistic(x,x0=3))
plt.title('$x_0=3$')
plt.tight_layout()
```
